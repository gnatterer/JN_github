```python
import pandas as pd

# Load the CSV files
file_a = "Earth_2022Q4_2025Q4_1_raw_combined.csv"
file_b = "Earth_2024 Q4_raw_combined.csv"
file_c = "Earth_2025 Q1_raw_combined.csv"

df_a = pd.read_csv(file_a)
df_b = pd.read_csv(file_b)
df_c = pd.read_csv(file_c)

# Ensure numeric columns are properly typed
for col in ["Royalties (USD)", "Units"]:
    df_a[col] = pd.to_numeric(df_a[col], errors="coerce")
    df_b[col] = pd.to_numeric(df_b[col], errors="coerce")
    df_c[col] = pd.to_numeric(df_c[col], errors="coerce")

# --- Checks BEFORE ---
print("=== BEFORE ===")
print("Total Royalties (USD) in A:", df_a["Royalties (USD)"].sum())
print("Total Units in A:", df_a["Units"].sum())

df_a_luna = df_a[df_a["Platform"] == "Luna"]
print("Total Royalties (USD) for Luna in A:", df_a_luna["Royalties (USD)"].sum())
print("Total Units for Luna in A:", df_a_luna["Units"].sum())

# --- Step 1: Delete rows from A ---
mask_remove = (df_a["Platform"] == "Luna") & (df_a["Statement Quarter"].isin(["2024 Q4", "2025 Q1"]))
df_a = df_a[~mask_remove]

# --- Step 2: Add rows from B and C ---
rows_b = df_b[(df_b["Platform"] == "Luna") & (df_b["Statement Quarter"] == "2024 Q4")]
rows_c = df_c[(df_c["Platform"] == "Luna") & (df_c["Statement Quarter"] == "2025 Q1")]

df_a = pd.concat([df_a, rows_b, rows_c], ignore_index=True)

# --- Checks AFTER ---
print("\n=== AFTER ===")
print("Total Royalties (USD) in A:", df_a["Royalties (USD)"].sum())
print("Total Units in A:", df_a["Units"].sum())

df_a_luna_after = df_a[df_a["Platform"] == "Luna"]
print("Total Royalties (USD) for Luna in A:", df_a_luna_after["Royalties (USD)"].sum())
print("Total Units for Luna in A:", df_a_luna_after["Units"].sum())

# Optionally save the updated file A
df_a.to_csv("fileA_updated.csv", index=False)
```
