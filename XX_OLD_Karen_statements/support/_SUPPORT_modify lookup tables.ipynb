```python



import pandas as pd
from opencc import OpenCC
import os

#lookup_song = '../All labels combined/lookup_tables/old/lookup_song_old.csv'
lookup_song = '../../50 KM Group/Royalties/Statements/Karen/Earth/Combined statements/lookup_tables/Earth_Lookup_isrc_song_album.csv'

def readfile(file):
    df = pd.read_csv(file)
    #df = pd.read_excel(file, sheet_name='Data')
    print(f"The dataframe '{file}' has {df.shape[0]} rows and {len(df.columns)} columns.")
    return df

orig_column = "ISRC.Song.Album"
mod_column = "ISRC.Song.Album_MOD"
#outputfile = "Rock_lookup_ISRC_song_album_albumtype_mod.xlsx"
outputfile = "Earth_Lookup_isrc_song_album_mod.csv"
outputdirectory = '../../50 KM Group/Royalties/Statements/Karen/_output/'

# Initialize the converter
converter = OpenCC('s2t') 

df_lookup_song = readfile(lookup_song)
duplicates = df_lookup_song[df_lookup_song.duplicated(subset=[orig_column], keep=False)]
#print(duplicates)


df_lookup_song[orig_column] = df_lookup_song[orig_column].astype(str)
df_lookup_song[mod_column] = (df_lookup_song[orig_column]
            .str.replace(" ", '', regex=False)
            .str.replace("(", '', regex=False)
            .str.replace(")", '', regex=False)
            .str.replace("-", '', regex=False)
            .str.replace("/", '', regex=False)
            .str.replace('（', '', regex=False)
            .str.replace('）', '', regex=False)
            .str.replace('’', '', regex=False)
            .str.replace("'", '', regex=False)
            .str.lower())

df_lookup_song[mod_column] = df_lookup_song[mod_column].apply(converter.convert)

#duplicates = df_lookup_song[df_lookup_song.duplicated(subset=['Song_original_lower'], keep=False)]
#print(duplicates)
df_lookup_song = df_lookup_song.drop_duplicates(subset=mod_column, keep="first")

path = os.path.join(outputdirectory, outputfile)
df_lookup_song.to_csv(path, index=True)
```
