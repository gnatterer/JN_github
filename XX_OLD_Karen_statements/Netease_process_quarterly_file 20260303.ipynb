```python

import pandas as pd
import os 
import numpy as np

# INPUTS

root_directory = '../../50 KM Group/Royalties/Statements/Karen/Netease/'
outputdirectory = '../../50 KM Group/Royalties/Statements/Karen/_output/'

#contract = 'Main contract' 
contract = 'Contract 5 songs 431A202404002900'
#contract = 'Main contract (ending 20281130) 431A202512054618' 
statement = '2026 Q2'
month1 = '2026-04'
month2 = '2026-05'
month3 = '2026-06'

# end inputs

subdirectory = '/as sent by Netease/'

directory = root_directory + contract + '/' + statement + subdirectory 

outputfilename = f"{outputdirectory}{contract}_{statement}.xlsx"

def find_xls_files(directory):
    xls_files = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if '结算报表' in file:
                xls_files.append(os.path.join(root, file))
            elif 'Revenue' in file:
                xls_files.append(os.path.join(root, file))
    return xls_files 

def read_each_sheet(file,m1,m2,m3):
    excel_data = pd.ExcelFile(file)
    print(f"\nreading {file}, all sheets")
    data_frames = []
    for sheet_name in excel_data.sheet_names:
        df = excel_data.parse(sheet_name)
        print(f"Reading sheet: {sheet_name}, Rows: {df.shape[0]}, Columns: {df.shape[1]}")
        if m1 in sheet_name:
            df['月份']= m1
        elif m2 in sheet_name:
            df['月份']= m2
        elif m3 in sheet_name:
            df['月份']= m3
        if "免费" in sheet_name:
            df['Total streams and download']=df['总播放量']+df['总下载量']
            df['类型']='免费'
            df.rename(columns={'本月分成收益费用': '本月实际分成收益费用 - royalties'}, inplace=True)
            data_frames.append(df)
        elif "Free" in sheet_name:
            df['Total streams and download']=df['总播放量 Total Number of Online Stream']+df['总下载量 Total Number of Download']
            df['类型']='免费'
            df.rename(columns={'歌曲名 Track Title': '歌曲名'}, inplace=True)
            df.rename(columns={'专辑 Album Title': '专辑'}, inplace=True)
            df.rename(columns={'艺人 Artist Name': '艺人'}, inplace=True)
            df.rename(columns={'付费类型 Fee Type': '付费类型'}, inplace=True)
            df.rename(columns={'非激励广告-总播放量 Total Number of Online Stream_Non-Ad Incentive': '非激励广告-总播放量'}, inplace=True)
            df.rename(columns={'激励广告-总播放量 Total Number of Online Stream_Ad-Incentive': '激励广告-总播放量'}, inplace=True)
            df.rename(columns={'总播放量 Total Number of Online Stream': '总播放量'}, inplace=True)
            df.rename(columns={'非激励广告-总下载量 Total Number of Download_Non-Ad Incentive': '非激励广告-总下载量'}, inplace=True)
            df.rename(columns={'总下载量 Total Number of Download': '总下载量'}, inplace=True)
            df.rename(columns={'录音授权比例 Percentage Of Recording Authorization': '录音授权比例'}, inplace=True)
            df.rename(columns={'词授权比例 Percentage Of Lyrics Authorization': '词授权比例'}, inplace=True)
            df.rename(columns={'曲授权比例 Percentage Of Composition Authorization': '曲授权比例'}, inplace=True)
            df.rename(columns={'表演者授权比例 Percentage Of Performers Authorization': '表演者授权比例'}, inplace=True)
            df.rename(columns={'非激励广告-本月分成收益费用 Revenue by Revenue Share_Non-Ad Incentive': '非激励广告-本月分成收益费用'}, inplace=True)
            df.rename(columns={'激励广告-本月分成收益费用 Revenue by Revenue Share_Ad-Incentive': '激励广告-本月分成收益费用'}, inplace=True)
            df.rename(columns={'本月分成收益费用 Revenue by Revenue Share': '本月分成收益费用'}, inplace=True)
            df.rename(columns={'本月结算金额 Monthly Revenue': '本月实际分成收益费用 - royalties'}, inplace=True)
            data_frames.append(df)
        elif '付费2' in sheet_name:
            df['Total streams and download']= df['总播放量']+df['总下载量'] 
            df['类型']='付费'
            df.rename(columns={'本月实际分成收益费用': '本月实际分成收益费用 - royalties'}, inplace=True)
            data_frames.append(df)
        elif '订阅' in sheet_name:
            df['Total streams and download']= df['总播放量']+df['总下载量'] 
            df['类型']='付费'
            df.rename(columns={'本月实际分成收益费用': '本月实际分成收益费用 - royalties'}, inplace=True)
            data_frames.append(df)
        elif 'Subscription' in sheet_name:
            df['Total streams and download']= df['总播放量 Total Number of Online Stream']+df['总下载量 Total Number of Download'] 
            df['类型']='付费'
            df.rename(columns={'歌曲名 Track Title': '歌曲名'}, inplace=True)
            df.rename(columns={'专辑 Album Title': '专辑'}, inplace=True)
            df.rename(columns={'艺人 Artist Name': '艺人'}, inplace=True)
            df.rename(columns={'付费类型 Fee Type': '付费类型'}, inplace=True)
            df.rename(columns={'总播放量 Total Number of Online Stream': '总播放量'}, inplace=True)
            df.rename(columns={'总下载量 Total Number of Download': '总下载量'}, inplace=True)
            df.rename(columns={'录音授权比例 Percentage Of Recording Authorization': '录音授权比例'}, inplace=True)
            df.rename(columns={'词授权比例 Percentage Of Lyrics Authorization': '词授权比例'}, inplace=True)
            df.rename(columns={'曲授权比例 Percentage Of Composition Authorization': '曲授权比例'}, inplace=True)
            df.rename(columns={'表演者授权比例 Percentage Of Performers Authorization': '表演者授权比例'}, inplace=True)
            #df.rename(columns={'本月实际分成收益费用 Revenue by Actual Revenue Share': '本月实际分成收益费用'}, inplace=True)
            df.rename(columns={'本月结算金额 Monthly Revenue': '本月实际分成收益费用 - royalties'}, inplace=True)
            data_frames.append(df)
        elif '付费单曲' in sheet_name:
            df.rename(columns={'销售数量': 'Total streams and download'}, inplace=True)
            df['类型']='付费单曲'
            df.rename(columns={'本月实际销售收益费用': '本月实际分成收益费用 - royalties'}, inplace=True)
            data_frames.append(df)
        elif 'Single' in sheet_name:
            df.rename(columns={'销售数量 No. of Sales': 'Total streams and download'}, inplace=True)
            df['类型']='付费单曲'
            df.rename(columns={'歌曲名 Track Title': '歌曲名'}, inplace=True)
            df.rename(columns={'专辑 Album Title': '专辑'}, inplace=True)
            df.rename(columns={'艺人 Artist Name': '艺人'}, inplace=True)
            df.rename(columns={'付费类型 Fee Type': '付费类型'}, inplace=True)
            df.rename(columns={'销售单价 Unit Price': '销售单价'}, inplace=True)
            df.rename(columns={'销售数量 No. of Sales': '销售数量'}, inplace=True)
            df.rename(columns={'实际销售收入 Actual Sales Revenue': '实际销售收入'}, inplace=True)
            df.rename(columns={'录音授权比例 Percentage Of Recording Authorization': '录音授权比例'}, inplace=True)
            df.rename(columns={'词授权比例 Percentage Of Lyrics Authorization': '词授权比例'}, inplace=True)
            df.rename(columns={'曲授权比例 Percentage Of Composition Authorization': '曲授权比例'}, inplace=True)
            df.rename(columns={'表演者授权比例 Percentage Of Performers Authorization': '表演者授权比例'}, inplace=True)
            df.rename(columns={'本月实际销售收益费用 Revenue by Actual Sales': '本月实际分成收益费用 - royalties'}, inplace=True)
            data_frames.append(df)
        elif 'K歌' in sheet_name:
            df['类型']='K歌'
            df['Total streams and download']= df['曲使用量']
            df.rename(columns={'本月单价收益费用': '本月实际分成收益费用 - royalties'}, inplace=True)
            data_frames.append(df)
    df = pd.concat(data_frames, ignore_index=True)
    df = df[df['ISRC'] != '合计']
    df = df[df['ISRC'] != '合计 Total']
    df['月份'] = df['月份'].str.replace('-', '')
    print(f"\nThe combined DataFrame has {df.shape[0]} rows and {len(df.columns)} columns.")
    print(f"Royalty: {df['本月实际分成收益费用 - royalties'].sum()}. Units: {df['Total streams and download'].sum()}")
    return df


listoffiles = find_xls_files(directory)
for file in listoffiles:
    df = read_each_sheet(file, month1, month2, month3)
df['Statement Quarter']=statement
df['Contract']=contract

df["Sales_date"] = pd.to_datetime(df["月份"], format="%Y%m", errors="coerce")
df["Sales Month"] = df["Sales_date"].dt.strftime("%Y %m")
df["Sales Quarter"] = (
    df["Sales_date"].dt.to_period("Q").astype(str).str.replace("Q", " Q")
)

df =df.drop(columns=['Sales_date'])

df1 = pd.read_excel("../../50 KM Group/Royalties/Statements/Karen/Netease/Combined statements/lookup/columns_sort.xlsx")

col_name_col = df1.columns[0]
order_col = df1.columns[1]

# Sort df1 by the order column (ascending)
df1_sorted = df1.sort_values(by=order_col)

# Extract ordered column list
ordered_columns = df1_sorted[col_name_col].tolist()

# Create any missing columns in df
for col in ordered_columns:
    if col not in df.columns:
        df[col] = np.nan

# Reorder df exactly according to df1 order
df = df[ordered_columns]

df.to_excel(outputfilename, engine='openpyxl', index=True)
```
