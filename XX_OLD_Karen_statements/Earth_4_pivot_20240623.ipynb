```python

# this script creates a pivot table from the matched 2022 Q4 - 2025Q1 statement

import pandas as pd


file ='../../50 KM Group/Royalties/Statements/Karen/Earth/Combined statements/Earth_2022Q4_2025Q4_2_matched.csv'
outputfilename3 = '../../50 KM Group/Royalties/Statements/Karen/_output/Earth_2022Q4_2025Q4_3_matched_pivot.xlsx'

df = pd.read_csv(file,low_memory=False)
print(f"The dataframe of the file '{file}' has {df.shape[0]} rows and {len(df.columns)} columns.")
print(f"Total fee: {df['Royalties (USD)'].sum()}. Total units: {df['Units'].sum()}")

for column in df.columns:        
        print(column)

# Filter 
#df_merged = df   #no filter
df_merged = df[df['Platform'] == 'oppo'].iloc[:, :] 
df_merged = df_merged.dropna(axis=1, how='all')
df_merged.to_excel("../../50 KM Group/Royalties/Statements/Karen/_output/oppo.xls", engine='openpyxl', index=True)
print(f"The filtered dataframe has {df_merged.shape[0]} rows and {len(df_merged.columns)} columns.")
print(f"Total fee: {df['Royalties (USD)'].sum()}. Total units: {df['Units'].sum()}")


 
```

```python
# Create Pivot table

val1= 'Units'
val2='Royalties (USD)'
val3='Share MABB (CNY)'

val1crit = 'sum'
val2crit = 'sum'
val3crit = 'sum'
val4crit = 'count'

x1='Product'
x2='Country'
x3='Platform'
x4='Statement Quarter'
x5='Song'
x6='Album Type'
x7 = 'Content'
x8 = 'Album'
x9 = 'Sales Month'
x10 = 'Sales Quarter'
x11 = 'ISRC'

y1 = ''

sortby = val2

group_cols = [x4,x10,x3,x9,x5]
for col in group_cols:
    if col in df.columns:
        df[col] = df[col].fillna("Blank")

pivot = pd.pivot_table(df_merged, 
                        values=[val1,val2],
                        index=group_cols,
                        fill_value=0,
                        margins= False,
                        aggfunc="sum")

# Convert MultiIndex to columns
pivot = pivot.reset_index()

# Forward-fill grouping columns so blanks become repeated values
pivot[group_cols] = pivot[group_cols].ffill()


#pivot = pd.pivot_table(df_merged, 
#                        values=[val1,val2],
#                        #index=[x4,x7,x3,x10,x9,x5],
#                        index=[x4,x3,x2,x11,x5],
##                        columns= y1,
#                        fill_value=0,
#                        margins= False,
#                        #aggfunc={val1:val1crit, val2:val2crit})
#                        aggfunc={val1:val1crit,val2:val2crit})

# pivot = pivot.ffill(axis=1).bfill(axis=1)

pivot.to_excel(outputfilename3, engine='openpyxl', index=True)
pivot
```
