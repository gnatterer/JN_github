```python
import pandas as pd
import os

input1 = "../../50 KM Group/Royalties/Statements/Karen/Earth/Combined statements/Earth_2022Q4_2025Q2_1_raw_combined.csv"
outputfilename = "../../50 KM Group/Royalties/Statements/Karen/_output/Earth_2022Q4_2025Q2_1_raw_combined_update.csv"

# Read the CSV
df = pd.read_csv(input1)

# Dictionary of row index → replacement value for "Song" (only if empty)
song_replacements = {
    160394: "The Way You Make Me Feel",
    163998: "当你老了",
    1115446: "慢慢喜欢你",
    2986: "扶摇",
    2987: "扶摇",
    163564: "这世界那么多人",
    316788: "这世界那么多人",
    319443: "这世界那么多人",
    573693: "这世界那么多人",
    577735: "这世界那么多人",
    711878: "这世界那么多人",
    844799: "这世界那么多人",
}

# Apply "Song" replacements only if empty
for row_idx, new_value in song_replacements.items():
    if row_idx in df.index:  # ensure the index exists
        if pd.isna(df.at[row_idx, "Song"]) or str(df.at[row_idx, "Song"]).strip() == "":
            df.at[row_idx, "Song"] = new_value
            print(f'Row {row_idx} in "Song" was empty. Updated to "{new_value}".')
        else:
            print(f'Row {row_idx} in "Song" is not empty (value = {df.at[row_idx, "Song"]}). No change made.')
    else:
        print(f"Row {row_idx} does not exist in the DataFrame.")

# Overwrite "Album" for row 160394 unconditionally
if 160394 in df.index:
    df.at[160394, "Album"] = "The Voyage"
    print('Row 160394 in "Album" updated to "The Voyage".')
else:
    print("Row 160394 does not exist in the DataFrame.")

# Save the updated DataFrame
df.to_csv(outputfilename, index=False)

print(f"\n✅ Updated file written to: {outputfilename}")
```
