```python
import pandas as pd

# Load the combined file
combined_file = "../../50 KM Group/Royalties/Statements/Other artists/Sally and Lam/SallyLam_2022h2_2025H2.csv"
df = pd.read_csv(combined_file)

# List of product titles to keep
titles_to_keep = [
    "XIAO SA ZOU YI HUI",
    "CHUN FENG QIU YU",
    "DARE TO LOVE ME",
    "ZHU FU",
    "QING REN ZHI JI",
    "FRIEND OR LOVERS?",
    "CENG JING XIN TONG",
    "QIU QU QIU LAI",
    "SHU ZI REN SHENG",
    "XUAN ZE"
]

# Filter the DataFrame
sub_df = df[df["Product Title"].isin(titles_to_keep)]

# Save the subfile to Excel
sub_df.to_excel("../../50 KM Group/Royalties/Statements/Karen/_output/SallyLam_2022H2_2025H2_subfile.xlsx", index=False)

# Print confirmation and stats
print("Subfile created: subfile.xlsx")
print("Number of rows in subfile:", len(sub_df))
print("Columns in subfile:", list(sub_df.columns))

# Optional: show totals for Sale Units and Royalty Payable in the subfile
if "Sale Units" in sub_df.columns:
    print("Total Sale Units in subfile:", sub_df["Sale Units"].sum())
if "Royalty Payable" in sub_df.columns:
    print("Total Royalty Payable in subfile:", sub_df["Royalty Payable"].sum())
```
