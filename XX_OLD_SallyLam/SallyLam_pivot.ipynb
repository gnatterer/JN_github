```python


import pandas as pd

# Load the combined file
combined_file = "../../50 KM Group/Royalties/Statements/Other artists/Sally and Lam/SallyLam_2022h2_2025H2.csv"
df = pd.read_csv(combined_file)

# Ensure numeric columns are properly typed
for col in ["Sale Units", "Royalty Payable"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Replace missing values in all grouping columns with "Blank"
#group_cols = ["SourceFile", "Product Type", "Digital Service Provider(DSP)", "Product Title"]
group_cols = ["SourceFile", "Digital Service Provider(DSP)", "Product Title"]
for col in group_cols:
    if col in df.columns:
        df[col] = df[col].fillna("Blank")

# Build the pivot table
pivot = pd.pivot_table(
    df,
    values=["Sale Units", "Royalty Payable"],
    index=group_cols,
    aggfunc="sum",
    fill_value=0
)


# Convert MultiIndex to columns
pivot = pivot.reset_index()

# Forward-fill grouping columns so blanks become repeated values
pivot[group_cols] = pivot[group_cols].ffill()

# Display the pivot table
print("=== Pivot Table ===")
print(pivot)

# Optionally save to CSV for easier viewing
pivot.to_excel("../../50 KM Group/Royalties/Statements/Karen/_output/SallyLam_2022h2_2025H2_pivot.xlsx", engine='openpyxl', index=True)
#pivot.to_csv("../../50 KM Group/Royalties/Statements/Karen/_output/SallyLam_2022h2_2025H2_pivot.csv")

```
