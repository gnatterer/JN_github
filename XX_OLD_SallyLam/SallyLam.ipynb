```python
import os
import pandas as pd

# Folder containing the CSV files
folder = "data"

# Collect all CSV files in the folder
csv_files = [f for f in os.listdir(folder) if f.lower().endswith(".csv")]

dataframes = []
file_stats = []

for file in csv_files:
    path = os.path.join(folder, file)
    df = pd.read_csv(path)

    # Add filename as first column
    df.insert(0, "SourceFile", file)

    # Ensure numeric columns are properly typed
    for col in ["Sale Units", "Royalty Payable"]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    # Collect stats for this file
    stats = {
        "file": file,
        "num_columns": len(df.columns),
        "num_rows": len(df),
        "sale_units_sum": df["Sale Units"].sum() if "Sale Units" in df.columns else None,
        "royalty_sum": df["Royalty Payable"].sum() if "Royalty Payable" in df.columns else None,
    }
    file_stats.append(stats)

    dataframes.append(df)

# Combine all files, aligning columns
combined_df = pd.concat(dataframes, ignore_index=True)

# --- Final stats ---
print("=== Individual File Stats ===")
for s in file_stats:
    print(f"File: {s['file']}")
    print(f"  Columns: {s['num_columns']}")
    print(f"  Rows: {s['num_rows']}")
    print(f"  Sale Units Sum: {s['sale_units_sum']}")
    print(f"  Royalty Payable Sum: {s['royalty_sum']}")
    print()

print("=== Combined File Stats ===")
print("Final list of columns:", list(combined_df.columns))
print("Total number of columns:", len(combined_df.columns))
print("Total number of rows:", len(combined_df))
if "Sale Units" in combined_df.columns:
    print("Total Sale Units Sum:", combined_df["Sale Units"].sum())
if "Royalty Payable" in combined_df.columns:
    print("Total Royalty Payable Sum:", combined_df["Royalty Payable"].sum())

# Optionally save combined file
combined_df.to_csv("combined.csv", index=False)
```
